```python
years = [2000, 2005, 2010, 2015]

for year in years:

    male_year = male_lit[male_lit['Year'] == year]
    female_year = female_lit[female_lit['Year'] == year]

    merged = male_year.merge(
        female_year,
        on='Country Code'
    )

    merged['Gender_Gap'] = merged['Value_x'] - merged['Value_y']

    avg_gap = merged['Gender_Gap'].mean()

    print(f"Year {year}: Average Gender Gap = {avg_gap}")
```
